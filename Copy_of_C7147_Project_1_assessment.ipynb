## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """Validate individual read characteristics (sequence is same length as the reference sequence (200 nt), only has valid characters (A/C/G/T) as bases).
    Repair otherwise valid reads with lower-case bases by converting to uppercase.
    Return a dictionary of read validation(uppercase sequence, whether it's valid, error of invalid read(wrong length or invalid character)). """
    # Write your code here.
    sequence = sequence.upper() #correct lowercase bases
    valid = True
    errors = []
    if len(sequence) != len(reference): #check length matching
        valid = False
        errors.append("wrong_length")
    if any(base not in "ACGT" for base in sequence): #check base validity
        valid = False
        errors.append("invalid_characters")
    return {"sequence": sequence, "valid": valid, "errors": errors} #build the dictionary
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """Identify the differences between a reference sequence and a read sequence, and the position of the mismatches."""
    # Write your code here.
    mutations_dict = []
    for i, base in enumerate(sequence): #number positions of bases
        if base != reference[i]: #check if bases at same position in reference and sequence are the same
            mutations_dict.append({"position": i + 1, #positioning starts from 1 not 0
                                   "reference_base": reference[i], #build the list of dictionaries
                                   "read_base": base})
    return mutations_dict
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Analyse reads and return their IDs, sample (1-6), group (control or treatment), sequences, validation status, errors and mutations."""
    # Write your code here.
    read_analysis_dictlist = []
    for read in reads:  # get read id, sample, group and sequence from reads list of dictionaries
        read_id = read["read_id"]
        sample = read["sample"]
        group = read["group"]
        sequence = read["sequence"]
        validated = validate_read(sequence, reference) #get validity and error info using validate_read
        sequence = validated["sequence"]
        if validated["valid"] == True:
            status = "valid"
            mutations = find_mutations(sequence, reference) #get mismatch info using find_mutations
        else:
            status = "invalid"
            mutations = []
        errors = validated["errors"]
        read_analysis_dictlist.append({  #build new list of dictionaries
            "read_id": read_id,
            "sample": sample,
            "group": group,
            "sequence": sequence,
            "status": status,
            "errors": errors,
            "mutations": mutations
        })
    return read_analysis_dictlist
read_analysis_dictlist = analyse_reads(reads, reference)
````

````python
# Write your code here.

def summarise_samples(analysed_reads):
  """Produce summary tables of of mismatch patterns between samples. Table 1: Summarised results including number of valid reads,
number of invalid reads, percentage of valid reads containing at least one mismatch, total number of mismatch events,
three positions with most mismatch events (fewer if less found), selecting earlier position in case of tie.
Table 2: Substitution mutations info with samples on columns and number of occurrences and percentage (per 100 valid reads) of
each observed nucleotide substitution on rows. """
  if not analysed_reads:
        return {}
  samples = {}    # group reads by sample
  for read in analysed_reads:
        sample = read["sample"]
        if sample not in samples:
            samples[sample] = []
        samples[sample].append(read)
  results = {}    # calculate statistics for each sample
  for sample, reads in samples.items():
        group = reads[0]["group"]
        valid_reads = 0
        invalid_reads = 0
        mismatch_reads = 0
        mismatch_events = 0
        substitution_counts = {}
        position_counts = {}
        for read in reads:
            if read["status"] == "invalid":  #count invalid reads
                invalid_reads += 1
            else:
                valid_reads += 1   #count valid reads
                mutations = read["mutations"]
                if mutations:
                    mismatch_reads += 1   #count mismatch reads and events
                    mismatch_events += len(mutations)
                    for mutation in mutations:   #count substitution types
                        substitution = (mutation["reference_base"] + ">" +
                                        mutation["read_base"])
                        if substitution not in substitution_counts:
                            substitution_counts[substitution] = 0
                        substitution_counts[substitution] += 1
                        position = mutation["position"]   #count mismatch positions
                        if position not in position_counts:
                            position_counts[position] = 0
                        position_counts[position] += 1
        if valid_reads > 0:   #calculate mismatch percentage
            mismatch_percentage = round((mismatch_reads / valid_reads) * 100, 2)
        else:
            mismatch_percentage = None
        sorted_positions = sorted(position_counts,   #find 3 most common mismatch positions
                                  key=lambda x: (-position_counts[x], x))
        top_positions = sorted_positions[:3]
        results[sample] = {"group": group,   #store sample results
            "valid_reads": valid_reads,
            "invalid_reads": invalid_reads,
            "mismatch_reads": mismatch_reads,
            "mismatch_percentage": mismatch_percentage,
            "mismatch_events": mismatch_events,
            "substitution_counts": substitution_counts,
            "position_counts": position_counts,
            "top_positions": top_positions}
  groups = {}    # group samples by experimental group
  for sample, stats in results.items():
        group = stats["group"]
        if group not in groups:
            groups[group] = []
        groups[group].append(sample)
    # Print results separately for each group
  for group, group_samples in groups.items():
        print(f"\nGROUP: {group}")
        # Table 1: Summarised results
        print("\nTABLE 1: Summarised results") #headings
        print(f"{'Sample':<12}{'Valid reads':<15}{'Invalid reads':<16}"
              f"{'Mismatch %':<15}{'Mismatch events':<18}{'Top positions'}")
        for sample in group_samples:  #information for table
            stats = results[sample]
            percentage = (f"{stats['mismatch_percentage']}%"
                          if stats["mismatch_percentage"] is not None
                          else "N/A")
            positions = ", ".join(str(position) for position in stats["top_positions"])
            print(f"{sample:<12}{stats['valid_reads']:<15}"
                  f"{stats['invalid_reads']:<16}{percentage:<15}"
                  f"{stats['mismatch_events']:<18}{positions}")
        # Table 2: Substitution mutations
        print("\nTABLE 2: Substitution mutations \n count (per 100 valid reads)")
        all_substitutions = []
        for sample in group_samples:  #get substitution information
            for substitution in results[sample]["substitution_counts"]:
                if substitution not in all_substitutions:
                    all_substitutions.append(substitution)
        all_substitutions.sort()  #sort substitution order
        print(f"{'Substitution':<18}", end="")
        for sample in group_samples:
            print(f"{sample:<22}", end="")
        print()
        for substitution in all_substitutions:
            print(f"{substitution:<18}", end="")
            for sample in group_samples:   #get substitution counts
                stats = results[sample]
                count = stats["substitution_counts"].get(substitution, 0)
                if stats["valid_reads"] > 0:   #get substitution percentages
                    percentage = (count / stats["valid_reads"]) * 100
                    value = f"{count} ({percentage:.2f}%)"
                else:
                    value = f"{count} (N/A)"
                print(f"{value:<22}", end="")
            print()

  return results
````

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
# Write your code here.

def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    """Compare mismatch percentages at the three most frequently mutated reference positions between control and treatment groups.
    Uses a two-sided Welch's t-test via calculate_statistics(). Returns list of dictionaries containing position-level statistics."""
    control_samples = {sample:data for sample,data in sample_summary.items() if data["group"] == control_group} #samples separated by group
    treated_samples = {sample:data for sample,data in sample_summary.items() if data["group"] == treated_group}
    all_samples = [data for data in list(control_samples.values()) + list(treated_samples.values()) if data["valid_reads"] > 0]
    if not all_samples:  #report groups with no samples
        print("No samples in groups.")
        return []
    combined_counts = {}  #sum all mismatch counts at same position from all samples
    for data in all_samples:
        for position, count in data["position_counts"].items():
            combined_counts[position] = (combined_counts.get(position, 0) + count)
    selected_positions = sorted([pos for pos, count in combined_counts.items() if count > 0], #get 3 top mismatch count positions
        key=lambda pos: (-combined_counts[pos], pos))[:3]     #in tie-break event, pick earlier reference position
    if not selected_positions:  #report groups with no mismatches
        print("No mismatches found in either group.")
        return []
    results = []
    for position in selected_positions:  #analyse each position
        control_percentages = []
        treated_percentages = []
        for sample_group, percentages in [(control_samples, control_percentages),   #calculate sample percentages
             (treated_samples, treated_percentages)]:
            for sample, data in sample_group.items():
                total_reads = data["valid_reads"]
                if total_reads == 0:  #exclude samples with no valid reads
                    print(f"Excluding sample '{sample}' "
                          f"from group '{data['group']}': no valid reads.")
                    continue
                mismatch_count = data["position_counts"].get(position, 0)  #missing positions don't contribute to mismatch count
                percentage = (mismatch_count / total_reads) * 100
                percentages.append(float(percentage))
        control_mean = (float(np.mean(control_percentages)) if control_percentages else None)  #calculate group means
        treated_mean = (float(np.mean(treated_percentages)) if treated_percentages else None)
        difference = (treated_mean - control_mean if control_mean is not None and treated_mean is not None else None)  #percentage difference
        statistical_result = calculate_statistics(control_percentages, treated_percentages) #statistical comparison using function
        statistic = statistical_result["statistic"]
        p_value = statistical_result["p_value"]
        results.append({"position": position,    #store results
            "control_n": len(control_percentages),
            "treated_n": len(treated_percentages),
            "control_mean": control_mean,
            "treated_mean": treated_mean,
            "difference": difference,
            "statistic": statistic,
            "p_value": p_value})
        if p_value is not None:   #print results in scientific style
            print(f"Position {position}: "
                f"{control_group} = {control_mean:.3f}% "
                f"(n = {len(control_percentages)}), "
                f"{treated_group} = {treated_mean:.3f}% "
                f"(n = {len(treated_percentages)}); "
                f"difference = {difference:.3f} percentage points; "
                f"Welch's t-test: t = {statistic:.3f}, "
                f"p = {p_value:.3g}.")
        else:
            print(f"Position {position}: "
                f"{control_group} = "
                f"{control_mean if control_mean is not None else 'NA'}%, "
                f"{treated_group} = "
                f"{treated_mean if treated_mean is not None else 'NA'}%; "
                f"statistical comparison unavailable "
                f"(n = {len(control_percentages)} vs "
                f"{len(treated_percentages)}).")
    return results
````
